#### **특이사항**
- Centre라는 지역을 2003년 disease 데이터에서 Centre (sans Ouagadougou), Ouagadougou로 나눠서 집계
- Ouagadougou가 주요 지역이라서 이렇게 집계한 듯
- 경기도라는 지역에 대해서 경기도, 경기도(수원시 제외), 수원시로 나누어 집계한 것과 같은 상황
- Ouagadougou의 인구(), Centre (sans Ouagadougou)의 인구()를 고려하여 하나의 Centre 발병률 행 생성해야 함

- ** Hauts Basins와 Hauts Bassins는 오타라고 판단하여 두 지역 모두 Hauts-bassins로 매핑했음**
  - 두 연도가 겹친다면 다뤄야할테지만, 두 집계는 각각 2010, 2016에 대한 집계라서 수정해줄 필요가 없음


###2006년 인구 데이터 기반 지역 내 인구 비율 계산:

- 'Ouagadougou'가 'Centre' 지역에서 차지하는 인구 비율: 1,181,702 / 1,727,390 = 약 68.4%

- 'Centre (sans Ouagadougou)'가 차지하는 인구 비율: 100% - 68.4% = 약 31.6%

### 2003년 추정치에 비율 적용:

2003년 'Centre' 지역 총 추정 인구 1,160,000명에 이 비율을 적용합니다.

####📋 2003년 최종 추정 인구
- 'Ouagadougou' (2003년 추정 인구): 1,160,000명 * 68.4% = 약 793,440명

- 'Centre (sans Ouagadougou)' (2003년 추정 인구): 1,160,000명 * 31.6% = 약 366,560명

- 가중 평균(Weighted Average)을 계산하실 때, Ouagadougou에는 약 79.3만 명, Centre (sans Ouagadougou)에는 약 36.7만 명의 인구 가중치를 적용하시면 됩니다.

```python
# 1. 정의: 상수 및 인구 정보 설정
COUNTRY = 'Burkina Faso'
YEAR = 2003
REGION_A = 'Ouagadougou'
REGION_B = 'Centre (sans Ouagadougou)'
NEW_REGION = 'Centre'

# 2003년 추정 인구 (이전 대화 기반)
POP_A = 793440  # Ouagadougou 2003 pop
POP_B = 366560  # Centre (sans...) 2003 pop
TOTAL_POP = POP_A + POP_B

# (주의) 이 코드는 disease_df가 메모리에 로드되어 있다고 가정합니다.
try:
    # 2. 데이터 추출: 2003년 Burkina Faso의 두 지역 데이터 필터링
    row_A = disease_df[
        (disease_df['country'] == COUNTRY) &
        (disease_df['year'] == YEAR) &
        (disease_df['region_name'] == REGION_A)
    ]
    row_B = disease_df[
        (disease_df['country'] == COUNTRY) &
        (disease_df['year'] == YEAR) &
        (disease_df['region_name'] == REGION_B)
    ]

    # 3. 유효성 검사: 2003년에 두 지역의 데이터가 모두 존재하는지 확인
    if row_A.empty or row_B.empty:
        print(f"오류: {YEAR}년 {COUNTRY}의 '{REGION_A}' 또는 '{REGION_B}' 데이터가 없습니다.")
        print("가중 평균을 계산할 수 없습니다.")
    else:
        # 4. 가중 평균 계산 (이미지의 'infection_rate' 컬럼 사용)
        rate_A = row_A['infection_rate'].iloc[0]
        rate_B = row_B['infection_rate'].iloc[0]

        # 분자: (비율 * 인구)의 합 = 총 발병 건수
        total_cases = (rate_A * POP_A) + (rate_B * POP_B)

        # 새로운 가중 평균 비율 계산
        new_weighted_rate = total_cases / TOTAL_POP

        # 5. 새로운 'Centre' 행 생성 (4개 주요 컬럼만)
        new_row = {
            'country': COUNTRY,
            'year': YEAR,
            'region_name': NEW_REGION,
            'infection_rate': new_weighted_rate
        }
        # (만약 4개 외 다른 컬럼이 있다면, disease_df.columns를 기반으로
        #  new_row를 채우는 더 복잡한 로직이 필요하지만,
        #  일반적으로는 주요 값만 채우고 나머지는 NaN으로 두어도 됩니다.)
        new_row_df = pd.DataFrame([new_row])

        # 6. 기존의 비표준 행(A, B) 삭제
        indices_to_drop = row_A.index.union(row_B.index)
        disease_df_cleaned = disease_df.drop(indices_to_drop)

        # 7. 새로운 'Centre' 행 추가 (원본 DataFrame에 다시 할당)
        # (중요) 이 줄은 disease_df 변수 자체를 업데이트합니다.
        disease_df = pd.concat([disease_df_cleaned, new_row_df], ignore_index=True)

        # 8. 결과 보고
        print(f"--- 작업 완료 (Year {YEAR}, {COUNTRY}) ---")
        print(f"'{REGION_A}' (Rate: {rate_A}) 및 '{REGION_B}' (Rate: {rate_B}) 행이 삭제되었습니다.")
        print(f"새로운 '{NEW_REGION}' 행이 가중 평균 Rate ({new_weighted_rate:.6f})로 생성되었습니다.")

        # (검증) 해당 연도/국가의 최종 데이터 출력
        print("\n--- 검증: 2003년 Burkina Faso 데이터 ---")
        print(disease_df[
            (disease_df['country'] == COUNTRY) &
            (disease_df['year'] == YEAR)
        ])

except NameError:
    print("오류: 'disease_df' DataFrame이 메모리에 정의되지 않았습니다.")
    print("이 코드를 실행하기 전에 disease_df를 로드해야 합니다.")
except KeyError:
    print("오류: 'infection_rate' 컬럼을 찾을 수 없습니다. 컬럼 이름을 확인하세요.")
except Exception as e:
    print(f"예상치 못한 오류 발생: {e}")
```

```python
# disease 데이터에서의 region name
print('disease region name')
print(sorted(set(disease_df[disease_df['country']=='Burkina Faso']['region_name'])))
# standard 데이터에서의 region name
print('\nstandard region name')
print(sorted(set(standard_df[standard_df['ADM0_NAME']=='Burkina Faso']['ADM1_NAME'])))


# disease와 standard에서 불일치하는 region name(disease 이름 기준)
print('\n\n\nmismatch list')
print(sorted(set(mismatch_df[mismatch_df['country']=='Burkina Faso']['region_name'])))

# 'Boucle de Mouhoun' : 'Boucle Du Mouhoun'
# 'Centre (sans Ouagadougou)' : ?
# 'Centre-Est' : 'Centre-est'
# 'Centre-Nord' : 'Centre-nord'
# 'Centre-Ouest' : 'Centre-ouest'
# 'Centre-Sud' : 'Centre-sud'
# 'Hauts Basins' : 'Hauts-bassins'
# 'Hauts Bassins' : 'Hauts-bassins'
# 'Ouagadougou' : ?
# 'Sud-Ouest' : 'Sud-ouest'



# 딕셔너리 만들기
burkina_mapping_dict = {
    'Boucle de Mouhoun' : 'Boucle Du Mouhoun',
    'Centre-Est' : 'Centre-est',
    'Centre-Nord' : 'Centre-nord',
    'Centre-Ouest' : 'Centre-ouest',
    'Centre-Sud' : 'Centre-sud',
    'Hauts Basins' : 'Hauts-bassins',
    'Hauts Bassins' : 'Hauts-bassins',
    'Sud-Ouest' : 'Sud-ouest'
}

disease_df.loc[disease_df['country'] == 'Burkina Faso', 'region_name'] = disease_df.loc[disease_df['country'] == 'Burkina Faso', 'region_name'].replace(burkina_mapping_dict)

print("disease data after mapping")
new_burkina_df = disease_df[disease_df['country']=='Burkina Faso']
print(new_burkina_df)

set(new_burkina_df['region_name']) - set(standard_df[standard_df['ADM0_NAME']=='Burkina Faso']['ADM1_NAME'])
```

```python
# 1. 연도별 인구 딕셔너리 정의
population_data = {
    2004: {
        'Yaounde': 1817524,
        'Centre(without Yaounde)': 1280520,
        'Douala': 1907479,
        'Littoral(without Douala)': 602784
    },
    2011: {
        'Yaounde': 2440400,
        'Centre(without Yaounde)': 1479400,
        'Douala': 2447100,
        'Littoral(without Douala)': 727800
    },
    2018: {
        'Yaounde': 3672800,
        'Centre(without Yaounde)': 1451000,
        'Douala': 3539200,
        'Littoral(without Douala)': 618900
    }
}

# 1. 연도별 딕셔너리를 하나의 '마스터 딕셔너리'로 묶습니다.
#    (Key: 연도, Value: 해당 연도의 매핑 딕셔너리)
master_mapping = {
    2004: {
        'Centre': 'Centre(without Yaounde)',
        'Littoral': 'Littoral(without Douala)'
    },
    2011: {
        'Centre': 'Centre(without Yaounde)',
        'Littoral': 'Littoral(without Douala)',
        'Yaoundé' : 'Yaounde'
    },
    2018: {
        'Centre (without Yaounde)|Centre (sans Yaoundé)' :'Centre(without Yaounde)',
        'Littoral (without Douala)' : 'Littoral(without Douala)'
    }
}

years_to_rename = [2004, 2011, 2018]

print("레이블 통일 시작...")
for year in years_to_rename:
    # 2. 현재 연도(year)에 맞는 딕셔너리를 마스터에서 꺼냅니다.
    current_mapping_dict = master_mapping[year]

    mask = (disease_df['country'] == 'Cameroon') & (disease_df['year'] == year)

    # 3. 꺼내온 'current_mapping_dict' (변수)를 replace에 적용합니다.
    disease_df.loc[mask, 'region_name'] = disease_df.loc[mask, 'region_name'].replace(current_mapping_dict)

print(f"{years_to_rename}년도 레이블 통일 완료.")


# 2. 집계할 연도 목록
years_to_process = [2004, 2011, 2018]
new_rows = [] # 합쳐진 행들을 임시 저장할 리스트

for year in years_to_process:
    print(f"\n--- {year}년 데이터 집계 시작 ---")
    pop = population_data[year]

    # --- A. Centre / Yaounde 합치기 ---
    regions_centre = ['Centre(without Yaounde)', 'Yaounde']
    df_centre_parts = disease_df[
        (disease_df['country'] == 'Cameroon') &
        (disease_df['year'] == year) &
        (disease_df['region_name'].isin(regions_centre))
    ]

    if len(df_centre_parts) == 2:
        rate_A = df_centre_parts.loc[df_centre_parts['region_name'] == 'Centre(without Yaounde)', 'infection_rate'].iloc[0]
        pop_A = pop['Centre(without Yaounde)']

        rate_B = df_centre_parts.loc[df_centre_parts['region_name'] == 'Yaounde', 'infection_rate'].iloc[0]
        pop_B = pop['Yaounde']

        # 가중 평균 계산
        total_cases = (rate_A * pop_A) + (rate_B * pop_B)
        total_pop = pop_A + pop_B
        new_rate = total_cases / total_pop

        new_rows.append({'country': 'Cameroon', 'year': year, 'region_name': 'Centre', 'infection_rate': new_rate})
        print(f"'{year} Centre' 생성 완료 (Rate: {new_rate:.6f})")
    else:
        print(f"경고: {year}년 Centre/Yaounde 데이터가 2개가 아닙니다. 건너뜁니다.")

    # --- B. Littoral / Douala 합치기 ---
    regions_littoral = ['Littoral(without Douala)', 'Douala']
    df_littoral_parts = disease_df[
        (disease_df['country'] == 'Cameroon') &
        (disease_df['year'] == year) &
        (disease_df['region_name'].isin(regions_littoral))
    ]

    if len(df_littoral_parts) == 2:
        rate_A = df_littoral_parts.loc[df_littoral_parts['region_name'] == 'Littoral(without Douala)', 'infection_rate'].iloc[0]
        pop_A = pop['Littoral(without Douala)']

        rate_B = df_littoral_parts.loc[df_littoral_parts['region_name'] == 'Douala', 'infection_rate'].iloc[0]
        pop_B = pop['Douala']

        # 가중 평균 계산
        total_cases = (rate_A * pop_A) + (rate_B * pop_B)
        total_pop = pop_A + pop_B
        new_rate = total_cases / total_pop

        new_rows.append({'country': 'Cameroon', 'year': year, 'region_name': 'Littoral', 'infection_rate': new_rate})
        print(f"'{year} Littoral' 생성 완료 (Rate: {new_rate:.6f})")
    else:
        print(f"경고: {year}년 Littoral/Douala 데이터가 2개가 아닙니다. 건너뜁니다.")


# 3. 기존의 비표준 행들 모두 삭제
regions_to_delete = [
    'Centre(without Yaounde)', 'Yaounde',
    'Littoral(without Douala)', 'Douala'
]
disease_df = disease_df[
    ~( (disease_df['country'] == 'Cameroon') &
       (disease_df['region_name'].isin(regions_to_delete)) )
]

# 4. 새로 생성된 'Centre', 'Littoral' 행들을 원본에 추가
if new_rows:
    disease_df = pd.concat([disease_df, pd.DataFrame(new_rows)], ignore_index=True)

print("\n--- [최종 작업 완료] ---")
print("카메룬 데이터 집계가 완료되었습니다.")
print(disease_df[disease_df['country'] == 'Cameroon'].sort_values(by='year'))
```

##**Zambia**

#### **mapping pair**
- Muchinga + Northern -(가중평균)-> Northern
- 'North Western' : 'North-Western'

#### **특이사항**

```python
# disease 데이터에서의 region name
print('disease region name')
print(sorted(set(disease_df[disease_df['country']=="Zambia"]['region_name'])))
# standard 데이터에서의 region name
print('\nstandard region name')
print(sorted(set(standard_df[standard_df['ADM0_NAME']=="Zambia"]['ADM1_NAME'])))


# disease와 standard에서 불일치하는 region name(disease 이름 기준)
print('\n\n\nmismatch list')
print(sorted(set(mismatch_df[mismatch_df['country']=="Zambia"]['region_name'])))
```

```python
# --- 1. 설정: 국가, 연도, 인구 ---
COUNTRY_NAME = 'Zambia'
YEARS_TO_PROCESS = [2013, 2018] # 처리할 연도 목록

# 2013년, 2018년 인구 (2010/2022년 인구 조사 기반 선형 보간 추정치)
population_data_by_year = {
    2013: {
        'Muchinga': 763317,
        'Northern': 1233971
    },
    2018: {
        'Muchinga': 849416,
        'Northern': 1447549
    }
}

print(f"--- 잠비아({COUNTRY_NAME}) [2013, 2018] 데이터 표준화 시작 ---")

# (오류 방지) disease_df가 존재하는지 확인
try:
    # disease_df의 존재 여부 확인 (데이터프레임인지)
    if not isinstance(disease_df, pd.DataFrame):
        print("오류: 'disease_df'가 DataFrame이 아닙니다. 먼저 로드해주세요.")
        raise NameError("'disease_df'가 정의되지 않음")

    # --- 2. 연도별(2013, 2018) 표준화 작업 반복 ---
    for year in YEARS_TO_PROCESS:

        print(f"\n--- {year}년 데이터 처리 시작 ---")

        # 해당 연도의 인구 데이터 가져오기
        population_data = population_data_by_year.get(year)
        if population_data is None:
            print(f"1단계 경고: {year}년 인구 데이터가 없습니다. 이 연도를 건너뜁니다.")
            continue

        # --- 2-1. [2:1 집계] 'Northern' 가중 평균 ---
        # (Muchinga + Northern -> Northern)

        regions_to_aggregate = ['Muchinga', 'Northern']
        new_region_name = 'Northern' # Standard 이름

        # 해당 연도, 국가, 2개 지역 데이터 필터링
        df_parts = disease_df[
            (disease_df['country'] == COUNTRY_NAME) &
            (disease_df['year'] == year) &
            (disease_df['region_name'].isin(regions_to_aggregate))
        ]

        new_row_df = pd.DataFrame() # 새 행을 담을 빈 DataFrame

        if len(df_parts) == 2:
            # 2개 지역이 모두 존재하면 가중 평균 계산
            pop_A = population_data['Muchinga']
            rate_A = df_parts.loc[df_parts['region_name'] == 'Muchinga', 'infection_rate'].iloc[0]

            pop_B = population_data['Northern']
            rate_B = df_parts.loc[df_parts['region_name'] == 'Northern', 'infection_rate'].iloc[0]

            total_cases = (rate_A * pop_A) + (rate_B * pop_B)
            total_pop = pop_A + pop_B
            new_rate = total_cases / total_pop

            # 새 행 생성
            new_row = {
                'country': COUNTRY_NAME,
                'year': year,
                'region_name': new_region_name, # 'Northern'
                'infection_rate': new_rate
            }
            new_row_df = pd.DataFrame([new_row])
            print(f"1단계: '{year} {new_region_name}' 생성 완료 (Rate: {new_rate:.6f})")

        elif len(df_parts) == 0:
            print(f"1단계 정보: {year}년 집계 대상(Muchinga, Northern)이 없습니다. 건너뜁니다.")
        else:
            print(f"1단계 경고: {year}년 2개 권역(Muchinga, Northern) 중 1개만 존재합니다. 집계를 건너뜁니다.")

        # --- 2-2. 불필요한 원본 행 삭제 ---
        # (df_parts에 있던 'Muchinga', 'Northern' 행)
        indices_to_drop = df_parts.index

        if len(indices_to_drop) > 0:
            disease_df = disease_df.drop(indices_to_drop)
            print(f"2단계: {year}년 비표준 행 {len(indices_to_drop)}개 삭제 완료.")
        # (집계할 행이 없었으면 삭제도 하지 않음)

        # --- 2-3. 새 집계 행 추가 ---
        if not new_row_df.empty:
            disease_df = pd.concat([disease_df, new_row_df], ignore_index=True)

        # --- 2-4. [1:1 매핑] 이름 수정 ---
        # ('North Western' -> 'North-Western')
        condition = (
            (disease_df['country'] == COUNTRY_NAME) &
            (disease_df['year'] == year) &
            (disease_df['region_name'] == 'North Western')
        )

        num_replaced = len(disease_df.loc[condition])
        if num_replaced > 0:
            disease_df.loc[condition, 'region_name'] = 'North-Western'
            print(f"3단계: 1:1 매핑 'North Western' -> 'North-Western' {num_replaced}건 처리 완료.")
        else:
            print(f"3단계: 1:1 매핑 대상('North Western')이 없습니다.")

    # --- 3. (루프 종료 후) 최종 검증 ---
    print(f"\n--- [최종 작업 완료] 잠비아({COUNTRY_NAME}) 데이터 정리 완료 ---")

    print(f"\n--- 검증 (처리된 연도: {YEARS_TO_PROCESS}) ---")
    standard_zambia_set = set(['Central', 'Copperbelt', 'Eastern', 'Luapula', 'Lusaka', 'North-Western', 'Northern', 'Southern', 'Western'])

    # 처리된 연도들의 지역명만 가져오기
    disease_set_processed = set(disease_df[
        (disease_df['country'] == COUNTRY_NAME) &
        (disease_df['year'].isin(YEARS_TO_PROCESS))
    ]['region_name'])

    # 표준 셋(standard_zambia_set)에 없는 지역명이 있는지 확인
    verification_result = disease_set_processed - standard_zambia_set
    print(f"표준 셋에 없는 지역명 (비어 있어야 함): {verification_result if verification_result else '없음'}")

    if not verification_result:
        print("✅ 검증 통과: 모든 지역명이 표준과 일치합니다.")
    else:
        print(f"⚠️ 검증 실패: {verification_result} 지역명이 표준화되지 않았습니다.")


# --- 4. (try...except) 오류 처리 ---
except NameError:
    # 'disease_df'가 정의되지 않았을 때
    print("오류: 'disease_df'가 메모리에 없습니다. 먼저 로드해주세요.")
except KeyError as e:
    # 'population_data_by_year' 딕셔너리에 'Muchinga' 또는 'Northern' 키가 없을 때
    print(f"오류: 인구 데이터 딕셔너리에 필요한 키가 없습니다: {e}")
except IndexError:
    # .iloc[0]을 호출했으나 데이터가 없을 때 (df_parts 로직이 잘못되었을 경우)
    print(f"오류: {year}년 데이터 필터링 중 인덱싱 오류가 발생했습니다.")
except Exception as e:
    # 기타 모든 예외
    print(f"예상치 못한 오류 발생: {e}")
```
